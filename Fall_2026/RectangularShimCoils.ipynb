```python
import runpy, numpy as np, matplotlib.pyplot as plt;
%run "ArbitraryPointBFieldRectangularHelmholtz.py"
```

```python
L1 = 10*0.0254  #  in -> m
L2 = 10*0.0254  #  in -> m

z=np.linspace(-0.05,0.05,1000) #(m)

Bz = np.array([
    HelmholtzField(
        0,          # x position (m)
        0,          # y position (m)
        zi,         # z position (m) changes at every iteration
        L1,         # L1: inner rectangular half-dimension (m)
        L2,         # L2: inner rectangular half-dimension (m)
        1,          # I: current (A)
        0.07023,    # d: distance center closest coil layer (m) 
        10,         # num_layers: number of layers
        1,          # num_loops: number of loops per layer
        5.12e-4     # Rwire: wire radius/spacing parameter (m) 
    )[2]            # Select Bz from [Bx, By, Bz]
    for zi in z
])


grad_z=np.gradient(Bz,z)/100
print(f"Gradient at center = {grad_z[np.argmin(np.abs(z))]:.2f} G/cm")
print("Bz =", Bz[np.argmin(np.abs(z))],"G")

fig,ax=plt.subplots(1,2,figsize=(12,5))

ax[0].plot(z*100,Bz)
ax[0].set(xlabel="z (cm)",ylabel="Bz (Gauss)",title="B field VS z")
ax[0].grid()

ax[1].plot(z*100,grad_z)
ax[1].set(xlabel="z (cm)",ylabel="dB/dz (G/cm)",title="Gradient VS z")
ax[1].grid() 

plt.tight_layout() 
plt.show()
```

```python
x = np.linspace(-0.05, 0.05, 1000) #(m)

Bx = np.array([
    HelmholtzField(
        xi,         # x (m)
        0,          # y (m)
        0,          # z (m) 
        0.106807,   # R: radius of the innermost loop (m) 
        1,          # I: current (A)
        0.07023,    # d: distance center closest coil layer (m) 
        10,         # num_layers: number of layers
        1,          # num_loops: number of loops per layer
        5.12e-4     # Rwire: wire radius/spacing parameter (m) 
    )[0]            # Select Bz from [Bx, By, Bz]
    for xi in x
])

grad_x = np.gradient(Bx, x) / 100
print(f"Gradient at center = {grad_x[np.argmin(np.abs(x))]:.2f} G/cm")

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].plot(x * 100, Bx)
ax[0].set(xlabel="x (cm)",ylabel="Bx (Gauss)",title="B field VS x")
ax[0].grid()

ax[1].plot(x * 100, grad_x)
ax[1].set(xlabel="x (cm)",ylabel="dBx/dx (G/cm)",title="Gradient VS x")
ax[1].grid()

plt.tight_layout()
plt.show()
```
