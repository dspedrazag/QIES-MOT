```python
import runpy, numpy as np, matplotlib.pyplot as plt;
%run "ArbitraryPointBField.py"
```

```python
z=np.linspace(-0.05,0.05,1000) #(m)

Bz = np.array([
    antiHelmholtzField(
        0,          # x position (m)
        0,          # y position (m)
        zi,         # z position (m) changes at every iteration
        0.0889,     # R: radius of the innermost loop (m)
        165,        # I: current (A)
        0.03730625, # d: distance center closest coil layer (m)
        14,         # num_layers: number of layers
        8,          # num_loops: number of loops per layer
        0.0028      # Rwire: wire radius/spacing parameter (m)
    )[2]            # Select Bz from [Bx, By, Bz]
    for zi in z
])

grad_z=np.gradient(Bz,z)/100
print(f"Gradient at center = {grad_z[np.argmin(np.abs(z))]:.2f} G/cm")

fig,ax=plt.subplots(1,2,figsize=(12,5))

ax[0].plot(z*100,Bz)
ax[0].set(xlabel="z (cm)",ylabel="Bz (Gauss)",title="B field VS z") 
ax[0].grid()

ax[1].plot(z*100,grad_z) 
ax[1].set(xlabel="z (cm)",ylabel="dB/dz (G/cm)",title="Gradient VS z") 
ax[1].grid() 

plt.tight_layout(); 
plt.show()
```

```python
x = np.linspace(-0.05, 0.05, 1000) #(m)

Bx = np.array([
    antiHelmholtzField(
        xi,         # x (m) 
        0,          # y (m)
        0,          # z (m)
        0.0889,     # R (m)
        165,        # I (A)
        0.03730625, # d (m)
        14,         # num_layers
        8,          # num_loops
        0.0028      # Rwire (m)
    )[0]            # Select Bx from [Bx, By, Bz]
    for xi in x
])

grad_x = np.gradient(Bx, x) / 100
print(f"Gradient at center = {grad_x[np.argmin(np.abs(x))]:.2f} G/cm")

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

ax[0].plot(x * 100, Bx)
ax[0].set(xlabel="x (cm)",ylabel="Bx (Gauss)",title="B field VS x")
ax[0].grid()

ax[1].plot(x * 100, grad_x)
ax[1].set(xlabel="x (cm)",ylabel="dBx/dx (G/cm)",title="Gradient VS x")
ax[1].grid()

plt.tight_layout()
plt.show()
```

```python
z=np.linspace(-0.05,0.05,1000) #(m)

Bz = np.array([
    antiHelmholtzField(
        0,          # x (m)
        0,          # y (m)
        zi,         # z (m)
        0.0889,     # R (m)
        1,          # I (A)
        0.03730625, # d (m)
        14,         # num_layers
        8,          # num_loops
        0.0028      # Rwire (m)        
    )[2]            # Select Bz from [Bx, By, Bz]
    for zi in z
])

grad_z=np.gradient(Bz,z)/100
print(f"Gradient at center = {grad_z[np.argmin(np.abs(z))]:.2f} G/cm/A")

fig,ax=plt.subplots(1,2,figsize=(12,5))

ax[0].plot(z*100,Bz)
ax[0].set(xlabel="z (cm)",ylabel="Bz (Gauss)",title="B field VS z") 
ax[0].grid()

ax[1].plot(z*100,grad_z) 
ax[1].set(xlabel="z (cm)",ylabel="dB/dz (G/cm)",title="Gradient VS z") 
ax[1].grid() 

plt.tight_layout() 
plt.show()
```
