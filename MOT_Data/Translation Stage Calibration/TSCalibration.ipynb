```python
import pandas as pd
import numpy as np

data = pd.read_csv("TranslationStageCalibration.csv")

x = data["TS(u.a)"]
y = data["RS(cm)"]

m, b = np.polyfit(x, y, 1) # Linear Regration: y = m*x + b

R = m * 39+ b #Checking value

print(f"Slope m = {m}")
print(f"Intercept = {b}")

print(f"R_cm = {m} * stage_reading + {b}")
print (R)
```

```python
import matplotlib.pyplot as plt

y_fit = m * x + b

plt.scatter(x, y, label="Calibration data")
plt.plot(x, y_fit, label="Linear fit")

plt.xlabel("Translation stage reading (a.u.)")
plt.ylabel("Ruler position (cm)")
plt.title("Translation Stage Calibration")

plt.grid()
plt.legend()
plt.show()
```
